## Import Data

```python
# Import packages
import os              
import numpy as np

import pandas as pd
import geopandas as gpd

import xarray as xr   
import rioxarray as rioxr
```

```python
# Read in fire perimeter data
fp1 = os.path.join('data', 'Eaton_Perimeter_20250121.geojson')
eaton = gpd.read_file(fp1)
fp2 = os.path.join('data','Palisades_Perimeter_20250121.geojson')
palisades = gpd.read_file(fp2)
```

```python
# Open the landsat 8 satellite reflectance data
fp3 = os.path.join('data', 'landsat8-2025-02-23-palisades-eaton.nc')
landsat = xr.open_dataset(fp3)
landsat
```

### Fire Perimeter Data for Eaton and Palisades:

```python
# Examine CRS details of Eaton and Palisades fire perimeter data

for df, name in zip([eaton, palisades], ['Eaton','Palisades']):
    print(f" {name} CRS: {df.crs}")
    print(f" {name} Datum: {df.crs.datum}")
    print(f" {name} is geographic?: {df.crs.is_geographic}")
    print(f" {name} is projected?: {df.crs.is_projected}")
```

Both datasets use the CRS EPSG:4326 and the crs is geographic. 

### Landsat Data:
The dimensions of the data are x(2742 coordinates) y(1418 coordinates) and time? (1 coordinate). The coordinates for Y and X are coordinates of projection in meters.

There are 6 variables being measured - red, green, blue, nir08, sqir22, and spatialref. Each of the first 5 variable values can be measured as a 2742x1418 array. The spatialref variable only holds one value and stores the geospatial information for the dataset.
